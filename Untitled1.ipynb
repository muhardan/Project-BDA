```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import torch
import transformers
import cv2
import pyspark

print(f"PyTorch Version : {torch.__version__}")
print(f"PySpark Version : {pyspark.__version__}")
print("Semua library siap digunakan!")
```

```python
# Cell 2: Membuat Dummy Data Transaksi E-Commerce
data = {
    'Transaction_ID': ['TRX001', 'TRX002', 'TRX003', 'TRX004', 'TRX005'],
    'Category': ['Electronics', 'Fashion', 'Electronics', 'Home & Living', 'Fashion'],
    'Price': [1500000, 250000, 3200000, 450000, 180000],
    'Quantity': [1, 3, 1, 2, 4]
}

df = pd.DataFrame(data)
df['Total_Sales'] = df['Price'] * df['Quantity']

print("=== Ringkasan Data Penjualan E-Commerce ===")
df
```

```python
print("=== INFO DATA ===")
print(f"Jumlah transaksi : {len(df)}")
print(f"Jumlah kolom     : {len(df.columns)}")
print("\nTipe data:")
print(df.dtypes)

print("\nCek missing value:")
print(df.isnull().sum())
```

```python
print("=== STATISTIK PENJUALAN ===")

print(f"Total penjualan : Rp{df['Total_Sales'].sum():,.0f}")
print(f"Rata-rata       : Rp{df['Total_Sales'].mean():,.0f}")
print(f"Penjualan tertinggi : Rp{df['Total_Sales'].max():,.0f}")
print(f"Penjualan terendah  : Rp{df['Total_Sales'].min():,.0f}")
print(f"Total barang terjual: {df['Quantity'].sum()}")
```

```python
sales_category = (
    df.groupby('Category')['Total_Sales']
      .sum()
      .sort_values(ascending=False)
)

print("=== PENJUALAN PER KATEGORI ===")
display(sales_category)
```

```python
plt.figure(figsize=(8, 5))

sales_category.plot(kind='bar')

plt.title('Total Penjualan Berdasarkan Kategori')
plt.xlabel('Kategori')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
quantity_category = (
    df.groupby('Category')['Quantity']
      .sum()
      .sort_values(ascending=False)
)

print("=== JUMLAH BARANG TERJUAL PER KATEGORI ===")
display(quantity_category)
```

```python
best_category = quantity_category.idxmax()
best_quantity = quantity_category.max()

print(f"Kategori dengan jumlah barang terjual terbanyak: {best_category}")
print(f"Jumlah barang terjual: {best_quantity}")
```

```python
highest_transaction = df.loc[df['Total_Sales'].idxmax()]

print("=== TRANSAKSI TERBESAR ===")
display(highest_transaction.to_frame().T)
```

```python
print("=== INSIGHT DATA ===")

best_sales_category = sales_category.idxmax()
best_sales_value = sales_category.max()

print(f"1. Kategori dengan total penjualan tertinggi adalah {best_sales_category}.")
print(f"   Total penjualannya sebesar Rp{best_sales_value:,.0f}.")

print(f"2. Kategori dengan jumlah barang terjual terbanyak adalah {best_category}.")
print(f"   Total barang terjual sebanyak {best_quantity} unit.")

print(f"3. Total seluruh penjualan adalah Rp{df['Total_Sales'].sum():,.0f}.")
print(f"4. Total barang yang terjual adalah {df['Quantity'].sum()} unit.")
```
